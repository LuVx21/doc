```go
import (
    "fmt"
    "github.com/samber/lo"
)

%%
names := lo.Uniq([]string{"foo", "bar", "foo"})
fmt.Println(names)

lo.ForEach([]string{"hello", "world"}, func(x string, _ int) {
    println(x)
})

groups := lo.GroupBy([]int{0, 1, 2, 3, 4, 5}, func(i int) int {
    return i % 3
})
fmt.Println(groups)
```
