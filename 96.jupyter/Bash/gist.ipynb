```python
%%bash
# 实现大括号扩展
function parse() {
    input="$1"
    if [[ "$input" =~ ^([^{]*)\{([^}]*)}(.*)$ ]]; then
        prefix="${BASH_REMATCH[1]}"
        versions_part="${BASH_REMATCH[2]}"
        suffix="${BASH_REMATCH[3]}"

        IFS=',' read -ra versions <<< "$versions_part"

        # result=""
        for version in "${versions[@]}"; do
            echo "${prefix}${version}${suffix}"
            # 拼接
            # [ -n "$result" ] && result+=","
            # result+="${prefix}${version}${suffix}"
        done

        # echo "$result"
    else
        echo "$input"
    fi
}
parse "alpine:{20,21}"
```
